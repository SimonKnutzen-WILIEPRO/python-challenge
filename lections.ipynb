## Programmablauf 

Wie verarbeitet der Computer unseren Befehlscode? 

**Sequentieller Ablauf von Befehlen:**

- Jede Zeile beinhaltet einen Befehl, der nacheinander ausgeführt wird
- Der Computer führt die Befehle in der Reihenfolge aus, in der sie geschrieben sind 
- Der Computer macht exakt das, was wir ihm sagen. D.h. wir müssen sehr genau sein.

 **Beispiel:** 
Textausgabe auf den Bildschirm mit dem Befehl _print_. Der Inhalt zwischen den geschweiften Klammern wird ausgeben. Wenn wir dem print-Befehl Text zum ausgeben übergeben wollen, muss dieser in Anführungszeichen stehen.

```python
print("Hallo Welt")
print("Hallo Welt 2")
```

## Befehle

Python beherrscht von sich aus viele grundlegende Befehle wie z.B. print, was Text auf dem Bildschirm ausgibt, oder grundrechenarten wie Addition,
Subtraktion, Multiplikation und Division. Zahlen und Rechenoperationen, die kein Bestandteil von Text sind, sondern zum Rechnen benutzt werden sollen, werden ohne Anführungszeichen eingegeben.

```python
print(2 + 3) 
```

Zum Vergleich mit Anfürungszeichen:

```python
print("2 + 3")
```

Darüber hinaus gibt es viele weitere (komplexere) Befehle, die in Python verwendet werden können. Viele dieser Befehle stehen nicht direkt zur verfügung, sondern sind in sogenannten Bibliotheken (oder Modulen) enthalten, aus denen diese erst importiert werden müssen, um sie nutzen zu können. Viele dieser Bibliotheken sind bereits in Python enthalten, andere müssen separat installiert werden.


**Beispiel: Aktuelles Datum und Uhrzeit abrufen:**

Hierzu muss der Befehl now() aus dem datetime-Modul importiert werden.
Dieser gibt das aktuelle Datum und die aktuelle Uhrzeit zurück.

Das datetime-Modul ist eine Standardbibliothek in Python, und ist bereits in der Python-Installation enthalten. Diese Bibliothek ermöglicht es uns, mit Datums- und Zeitangaben zu arbeiten.

```python
from datetime import datetime
print(datetime.now()) 
```

**Befehlssyntax**

- Zur Syntax der meisten Befehle in Python gehört, dass sie mit einem Namen beginnen, gefolgt von Klammern, die Parameter enthalten können.
- Sollte der Befehl keine Parameter benötigen, können die Klammern leer bleiben.
- Da der now()-Befehl aus dem datetime-Modul stammt, muss er mit dem Modulnamen (datetime) und einem Punkt (.) aufgerufen werden.

## Variablen

Anstatt das Ergebnis eines Befehls sofort auszugeben (print), können wir es auch in einer Variablen speichern, um es später im Code weiterzuverwenden.

Variablen sind wie Kisten, in denen wir temporär Werte abspeichern können. In diese können wir nicht nur die Ergebnisse eines Befehls abspeichern, sondern wir können diesen auch direkt Werte zuweisen. Die Zuweisung eines Werte erfolgt über den = Operator.
Diese Variablen exisiteren jedoch nur, solange das Programm läuft. Wenn das Programm beendet wird, werden alle Variablen wieder gelöscht.

Der Wert einer Variablen kann mit dem print Befehl ausgegeben werden.

```python
ergebnis_addition = 2 + 3
print("ergebnis_addition: ")
print(ergebnis_addition)
substraktor = 1
ergebnis_final = ergebnis_addition - substraktor
print("ergebnis_final: ")
print(ergebnis_final)
```
