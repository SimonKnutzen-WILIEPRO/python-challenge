## Programmablauf 

Wie verarbeitet der Computer unseren Quellcode? 

**Sequentieller Ablauf von Befehlen:**

- Jede Zeile beinhaltet einen Befehl, der nacheinander ausgeführt wird
- Der Computer führt die Befehle in der Reihenfolge aus, in der sie geschrieben sind 
- Der Computer macht exakt das, was wir ihm sagen. D.h. wir müssen sehr genau sein.

 **Beispiel:** 
Textausgabe auf den Bildschirm mit dem Befehl _print_. Der Inhalt zwischen den geschweiften Klammern wird ausgeben. Wenn wir dem print-Befehl Text zum ausgeben übergeben wollen, muss dieser in Anführungszeichen stehen.

```python
print("Hallo Welt")
print("Hallo Welt 2")
```

## Befehle

Python beherrscht von sich aus viele grundlegende Befehle wie z.B. print, was Text auf dem Bildschirm ausgibt, oder grundrechenarten wie Addition,
Subtraktion, Multiplikation und Division. Zahlen und Rechenoperationen, die kein Bestandteil von Text sind, sondern zum Rechnen benutzt werden sollen, werden ohne Anführungszeichen eingegeben.

```python
print(2 + 3) 
```

Zum Vergleich mit Anfürungszeichen:

```python
print("2 + 3")
```

> Deep Dive: Die Anführungszeichen bewirken, dass die Zeichen, welche sich zwischen den Anführungszeichen befinden als ein sogenannter _string_ interpretiert werden. String ist einer von vielen Datentypen, die in Python unterstützt werden. Er wird verwendet um Text zu repräsentieren.

Darüber hinaus gibt es viele weitere (komplexere) Befehle, die in Python verwendet werden können. Viele dieser Befehle stehen nicht direkt zur verfügung, sondern sind in sogenannten Bibliotheken (oder Modulen) enthalten, aus denen diese erst importiert werden müssen, um sie nutzen zu können. Viele dieser Bibliotheken sind bereits in Python enthalten, andere müssen separat installiert werden.


**Beispiel: Aktuelles Datum und Uhrzeit abrufen:**

Hierzu muss der Befehl now() aus dem datetime-Modul importiert werden.
Dieser gibt das aktuelle Datum und die aktuelle Uhrzeit zurück.

Das datetime-Modul ist eine Standardbibliothek in Python, und ist bereits in der Python-Installation enthalten. Diese Bibliothek ermöglicht es uns, mit Datums- und Zeitangaben zu arbeiten.

```python
from datetime import datetime
print(datetime.now()) 
```

**Befehlssyntax**

- Zur Syntax der meisten Befehle in Python gehört, dass sie mit einem Namen beginnen, gefolgt von Klammern, die Parameter enthalten können.
- Sollte der Befehl keine Parameter benötigen, können die Klammern leer bleiben.
- Da der now()-Befehl aus dem datetime-Modul stammt, muss er mit dem Modulnamen (datetime) und einem Punkt (.) aufgerufen werden.

# Lektion 2 - Variablen und Datentypen

## Was sind Variablen?

Anstatt das Ergebnis eines Befehls sofort auszugeben (print), können wir es auch in einer Variablen speichern, um es später im Code weiterzuverwenden.

Variablen sind wie Kisten, in denen wir temporär Werte abspeichern können. In diese können wir nicht nur die Ergebnisse eines Befehls abspeichern, sondern wir können diesen auch direkt Werte zuweisen. Die Zuweisung eines Werte erfolgt über den = Operator.
Diese Variablen exisiteren jedoch nur, solange das Programm läuft. Wenn das Programm beendet wird, werden alle Variablen wieder gelöscht.

Der Wert einer Variablen kann mit dem print Befehl ausgegeben werden.

```python
ergebnis_addition = 2 + 3
print("ergebnis_addition: ")
print(ergebnis_addition)
substraktor = 1
ergebnis_final = ergebnis_addition - substraktor
print("ergebnis_final: ")
print(ergebnis_final)
```

## Datentypen von Variablen

Variablen können verschiedene Datentypen besitzen. 

Einen ersten Datentyp haben wir bereits in der ersten Lektion kennengelernt: _string_ (Dieser wird benutzt, um Text in Variablen abzuspeichern)

Der Datentyp einer Variablen muss in Python nicht explizit angegeben werden.
Python erkennt den Datentyp automatisch, basierend auf dem zugewiesenen Wert. 
Wird die Variable später mit einem anderen Datentyp überschrieben, ändert sich der Datentyp der Variable entsprechend.
Die beiden anderen wichtigen Datentypen, die wir in dieser Lektion kennenlernen, sind _integer_ (Ganzzahlen) und _float_ (Gleitkommazahlen).
Ein Integer ist eine ganze Zahl, z.B. 1, 2, 3, -1, -2, -3.
Ein Float ist eine Zahl mit Dezimalstellen, z.B. 1.0, 2.5, -3.14, 0.001.

Wenn wir eine Variable in einem Befehl verwenden, wird der Datentyp der Variable automatisch erkannt und entsprechend verarbeitet. 
Verwenden wir beispielsweise eine Variable, die einen String enthält, in einer mathematischen Operation, wird Python einen Fehler ausgeben.

```python
number = 2
text = "4"
print(number + text)
```

Die Fehlermeldung sagt uns, dass es nicht möglich ist eine Addition zwischen einem _string_ und einem _integer_ auszuführen

> Deep Dive: Vorsicht! Manche mathematischen Operatoren sind in Python "überladen". D.H. sie haben mehr als eine Funktionsweise, die von den Datentypen der verwendeten Variablen abhängt. So kann man einen _string_ mit einem _integer_ multiplizieren, doch das Ergebnis ist nicht unbedingt das, was man erwartet: In diesem Fall (_string_ * __integer_) wird die Zeichenkette des Strings mehrfach wiederholt (so oft, wie mit dem _integer_ angegeben)

```python
print("4" * 3)
```

Der letzte noch nicht erwähnte elementare Datentyp in Python ist der boolesche Datentyp.
Er kann nur zwei Werte annehmen: True oder False. Er wird verwendet, um den Wahrheitswert von Ausdrücken zu repräsentieren.



# Lektion 3 - Sammlungen 

## Listen

Neben den elementaren Datentypen gibt es in Python noch Datentypen, die Sammlungen von Werten repräsentieren. Der häufigste ist die Liste. Eine Variable dieses Datentyps kann eine Vielzahl von anderen Werten enthalten. Listen sind in Python sehr flexibel und können gleichzeitig verschiedene Datentypen enthalten, einschließlich anderer Listen.

Listen werden durch eckige Klammern gebildet, zwischen denen die verschiedenen Werte durch Komma getrennt angegeben werden.

```python
liste1 = [1, 2, 3]
liste2 = ["4", "5", "6"]
liste3 = ["Hallo", 2, 3.14]
liste4 = [1, "Hallo", 3.14, [2, 3, 4]]
```

```python
print(liste1)
print(liste2)
print(liste3)
print(liste4)
```

> Anmerkung: In den meisten Fällen ist es nicht sinnvoll, eine Liste mit gemischten Datentypen zu benutzen, jedoch gibt es besondere Anwendungsfälle, in denen diese Möglichkeit von Nutzen ist.

Um auf einen einzelnen Wert innerhalb der Liste zuzugreifen, benutzen wir den Index des Eintrages (die Position des Eintrages von vorne gezählt). 
Der Zugriff erfolgt, indem wir hinter den Namen der List eckige Klammern setzen und dazwischen den Index angeben.

Achtung! Python fängt bei 0 an zu zählen, d.h. der erste Eintrag der Liste hat den Index _0_

```python
print(liste1[1])
```

Es ist möglich, einer Liste weitere Elemente hinzuzufügen. Hierzu gibt es die Methode append(), die auf eine Liste angewendet wird und ein weiteres Element an das Ende der Liste anfügt

```python
liste1.append(4)
print(liste1)
```

Es ist ebenfalls möglich, einzelne Elemente zu überschreiben. Hierzu wird der Index des Elements verwendet, zusammen mit der Zuweisung eines neuen Wertes.


liste1[0] = 10
print(liste1)

Es ist ausserdem moeglich, Listen zu addieren. Hierbei werden die Elemente der Listen hintereinander angehaengt.

```python
liste5 = [4, 5, 6]
liste_combined = liste1 + liste5
print(liste_combined)
```

Man kann ausserdem eine Sub-Liste aus einer Liste extrahieren. Hierzu gibt es den sogenannten "Slicing"-Operator, bei welchem man den Start- und End-Index in eckigen Klammern hinter dem Listennamen angibt (getrennt durch einen Doppelpunkt)

```python
list_subset = liste_combined[2:4]
print(list_subset)
```

Der erste Index ist inklusiv, der zweite exklusiv. Um den letzten Index zu inkludieren, lassen wir den End-Index leer.

```python
list_subset2 = liste_combined[2:]
print(list_subset2)
```

Es gibt noch viele weitere Möglichkeiten, mit Listen zu arbeiten.
wie z.B. das Einfügen, Entfernen oder Sortieren von Elementen.
